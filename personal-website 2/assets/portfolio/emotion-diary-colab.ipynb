```python
%%writefile data_manager.py
import json
import os

def load_data(filename):
    file1 = os.path.exists(filename)
    if file1:
        file = open(filename, 'r')
        content = file.read()
        file.close()
        data = json.loads(content)
        return data
    else:
        empty = []
        return empty

def save_data(filename, data):
    jsons = json.dumps(data, indent=4)
    file = open(filename, 'w')
    file.write(jsons)
    file.close()


def add_record(filename, record):
    edata = load_data(filename)
    edata.append(record)
    save_data(filename, edata)
```

```python
%%writefile visualization.py
from datetime import datetime, timedelta
import random


def s_txt(a):
    b = []
    i = 0
    while i < len(a):
        b.append(a[i])
        i += 1
    b.sort()
    return b

def s_day(a):
    b = []
    i = 0
    while i < len(a):
        b.append(a[i])
        i += 1
    i = 0
    while i < len(b):
        j = i + 1
        while j < len(b):
            if b[j]['date'] < b[i]['date']:
                t = b[i]
                b[i] = b[j]
                b[j] = t
            j += 1
        i += 1
    return b

def get_weekly_summary(a):
    wk = {}
    for d in a:
        s=a[d]
        dt = datetime.strptime(d,'%Y-%m-%d')
        w=dt.strftime('%Y-W%U')
        if w not in wk:
            wk[w] = {}
            wk[w]['dates'] = []
            wk[w]['ints'] = []
            wk[w]['emos'] = {}
            wk[w]['cnt'] = 0
            wk[w]['tops'] = []
        g = wk[w]
        g['dates'].append(d)
        g['ints'].append(s['average_intensity'])
        g['cnt'] = g['cnt'] + s['count']
        if 'main_emotions' in s:
            ms = s['main_emotions']
            if ms is None:
                ms = []
            i = 0
            while i < len(ms):
                g['tops'].append(ms[i])
                i += 1
        if 'all_scores' in s:
            for e in s['all_scores']:
                sc = s['all_scores'][e]
                if e not in g['emos']:
                    g['emos'][e] = 0.0
                g['emos'][e] = g['emos'][e] + sc
    ws = []
    keys = []
    for k in wk:
        keys.append(k)
    keys.sort()
    keys.reverse()

    i=0
    while i < len(keys) and i < 12:
        k = keys[i]
        x = wk[k]
        tot = 0
        n = 0
        j = 0
        while j < len(x['ints']):
            tot = tot + x['ints'][j]
            n += 1
            j += 1
        avg = 0
        if n > 0:
            avg = tot / n
        dom = None
        ma = 0
        for e in x['emos']:
            if dom is None:
                dom = e
                ma = x['emos'][e]
            else:
                if x['emos'][e] > ma:
                    dom = e
                    ma = x['emos'][e]
        md=None
        if len(x['dates']) > 0:
            d0 = x['dates'][0]
            if d0 in a:
                if 'emotion_metadata' in a[d0]:
                    md = a[d0]['emotion_metadata']
                else:
                    md = {}

        en=None
        if dom is not None and md is not None:
            if dom in md:
                if 'encouragement' in md[dom]:
                    arr = md[dom]['encouragement']
                    if len(arr) > 0:
                        idx = random.randint(0, len(arr) - 1)
                        en = arr[idx]
        ds = s_txt(x['dates'])
        if len(ds) > 0:
            wtxt = ds[0] + " → " + ds[len(ds) - 1]
        else:
            wtxt = k

        emo_copy = {}
        for e in x['emos']:
            emo_copy[e] = x['emos'][e]
        row = {}
        row['week'] = wtxt
        row['week_key'] = k
        row['avg_intensity'] = avg
        row['dominant_emotion'] = dom
        row['total_entries'] = x['cnt']
        row['emotions'] = emo_copy
        row['dates'] = ds
        row['encouragement'] = en
        row['emotion_metadata'] = md
        ws.append(row)
        i += 1

    return ws

def get_monthly_summary(a):
    mm={}
    for d in a:
        s=a[d]
        m = d[:7]
        if m not in mm:
            mm[m] = {}
            mm[m]['dates'] = []
            mm[m]['ints'] = []
            mm[m]['emos'] = {}
            mm[m]['cnt'] = 0
            mm[m]['days'] = []

        p = mm[m]
        p['dates'].append(d)
        p['ints'].append(s['average_intensity'])
        p['cnt'] = p['cnt'] + s['count']

        x1 = {}
        x1['date'] = d
        x1['intensity'] = s['average_intensity']
        p['days'].append(x1)

        if 'all_scores' in s:
            for e in s['all_scores']:
                sc = s['all_scores'][e]
                if e not in p['emos']:
                    p['emos'][e] = 0.0
                p['emos'][e] = p['emos'][e] + sc

    out = []
    keys = []
    for k in mm:
        keys.append(k)
    keys.sort()
    keys.reverse()
    i=0
    while i < len(keys) and i < 6:
        k = keys[i]
        x = mm[k]
        avg=0
        mx=0
        mn=0
        if len(x['ints']) > 0:
            j = 0
            tot = 0
            mx = x['ints'][0]
            mn = x['ints'][0]
            while j < len(x['ints']):
                v = x['ints'][j]
                tot = tot + v
                if v > mx:
                    mx = v
                if v < mn:
                    mn = v
                j += 1
            avg = tot / len(x['ints'])
        dom = None
        ma = 0
        for e in x['emos']:
            if dom is None:
                dom = e
                ma = x['emos'][e]
            else:
                if x['emos'][e] > ma:
                    dom = e
                    ma = x['emos'][e]
        tr=0
        if len(x['days']) > 1:
            ds = s_day(x['days'])
            n = len(ds)

            xmean = 0
            ymean = 0
            j = 0
            while j < n:
                xmean = xmean + j
                ymean = ymean + ds[j]['intensity']
                j += 1
            xmean = xmean / n
            ymean = ymean / n
            up = 0
            down = 0
            j = 0
            while j < n:
                dx = j - xmean
                dy = ds[j]['intensity'] - ymean
                up = up + (dx * dy)
                down = down + (dx * dx)
                j += 1
            if down != 0:
                tr = up / down
        emo_copy = {}
        for e in x['emos']:
            emo_copy[e] = x['emos'][e]

        row = {}
        row['month'] = k
        row['avg_intensity'] = avg
        row['max_intensity'] = mx
        row['min_intensity'] = mn
        row['dominant_emotion'] = dom
        row['total_entries'] = x['cnt']
        row['total_days'] = len(x['dates'])
        row['emotions'] = emo_copy
        row['trend'] = tr
        row['daily_data'] = s_day(x['days'])
        out.append(row)

        i += 1
    return out

def hex_to_rgb(h):
    s = h.lstrip('#')
    r = int(s[0:2], 16)
    g = int(s[2:4], 16)
    b = int(s[4:6], 16)
    return (r, g, b)

def rgb_to_hex(x):
    r = int(x[0])
    g = int(x[1])
    b = int(x[2])

    if r < 0:
        r = 0
    if r > 255:
        r = 255
    if g < 0:
        g = 0
    if g > 255:
        g = 255
    if b < 0:
        b = 0
    if b > 255:
        b = 255
    rh = hex(r)[2:]
    gh = hex(g)[2:]
    bh = hex(b)[2:]
    if len(rh) == 1:
        rh = '0' + rh
    if len(gh) == 1:
        gh = '0' + gh
    if len(bh) == 1:
        bh = '0' + bh

    return '#' + rh + gh + bh

def adjust_saturation(h, lv, mn=1, mx=5):
    r, g, b = hex_to_rgb(h)

    p = 40 + (lv - mn) * (100 - 40) / (mx - mn)
    f = p / 100.0

    if f < 0.4:
        f = 0.4
    if f > 1.0:
        f = 1.0

    gray = (r + g + b) / 3
    r2 = gray + (r - gray) * f
    g2 = gray + (g - gray) * f
    b2 = gray + (b - gray) * f

    return rgb_to_hex((r2, g2, b2))

def mix_colors(cs, ws=None):
    if len(cs) == 0:
        return '#CCCCCC'

    if len(cs) == 1:
        return cs[0]

    if ws is None:
        ws = []
        i = 0
        while i < len(cs):
            ws.append(1.0)
            i += 1
    total = 0
    i = 0
    while i < len(ws):
        total = total + ws[i]
        i += 1

    if total == 0:
        total = 1
    i = 0
    while i < len(ws):
        ws[i] = ws[i] / total
        i += 1
    rgbs = []
    i = 0
    while i < len(cs):
        rgbs.append(hex_to_rgb(cs[i]))
        i += 1
    r = 0
    g = 0
    b = 0
    i = 0
    lim = len(rgbs)
    if len(ws) < lim:
        lim = len(ws)
    while i < lim:
        rgb = rgbs[i]
        w = ws[i]
        r = r + rgb[0] * w
        g = g + rgb[1] * w
        b = b + rgb[2] * w
        i += 1

    return rgb_to_hex((r, g, b))

def generate_calendar_data(a):
    cols = {}
    cols['joy'] = '#FDD835'
    cols['sadness'] = '#42A5F5'
    cols['anger'] = '#D32F2F'
    cols['fear'] = '#26C6DA'
    cols['disgust'] = '#9CCC65'
    cols['surprise'] = '#FF7043'

    out=[]
    for d in a:
        s=a[d]
        it = s['average_intensity']
        dm = s['dominant_emotion']

        if 'main_emotions' in s:
            ms = s['main_emotions']
        else:
            ms=[dm]

        if dm and dm in cols:
            base = cols[dm]
            c = adjust_saturation(base, it)
        else:
            c = '#CCCCCC'

        x1 = {}
        x1['date'] = d
        x1['color'] = c
        x1['emotion_colors'] = [c]
        x1['dominant_emotion'] = dm
        x1['main_emotions'] = ms
        x1['count'] = s['count']
        x1['intensity'] = it
        out.append(x1)

    return out
```

```python
%%writefile app.py
from flask import Flask, render_template, request, redirect
from data_manager import load_data, save_data, add_record
from models import analyze_emotions
from visualization import generate_calendar_data, get_weekly_summary, get_monthly_summary
from datetime import datetime
from calendar import monthrange
import random

app = Flask(__name__)
FILENAME = "data.json"

#input from index
@app.route("/")
def index():
    return render_template("index.html")

#uplaod
@app.route("/submit", methods=["POST"])
def submit():
    date =request.form["date"]
    text =request.form["text"]

    names= request.form.getlist("emotion_name[]")
    levels= request.form.getlist("emotion_intensity[]")
    emotions = {}
    i= 0
    while i < len(names):
        n = names[i]
        lv = levels[i]
        if len(n.strip()) > 0:
            emotions[n.strip()] = int(lv)
        i= i + 1

    tag_str = request.form.get("tags", "")
    tag_list = tag_str.split(",")
    tags = []
    for t in tag_list:
        t = t.strip()
        if len(t) > 0:
            tags.append(t)
    record= {}
    record["date"] =date
    record["text"] =text
    record["emotions"] =emotions
    record["tags"] =tags
    add_record(FILENAME, record)
    return redirect("/result")

@app.route("/result")
def result():
    data = load_data(FILENAME)
    analysis=analyze_emotions(data)
    return render_template("result.html", analysis=analysis)

@app.route("/calendar")
def calendar():
    data=load_data(FILENAME)
    analysis=analyze_emotions(data)
    cal_data =generate_calendar_data(analysis)
    by_date = {}
    for item in cal_data:
        by_date[item['date']] = item

    months = []
    for item in cal_data:
        m = item['date'][:7]
        if m not in months:
            months.append(m)
    months.sort()
    months.reverse()

    recent = []
    cnt =0
    for m in months:
        if cnt< 3:
            recent.append(m)
            cnt +=1

    result= {}
    for m in recent:
        parts =m.split('-')
        yr= int(parts[0])
        mo =int(parts[1])
        start, days= monthrange(yr, mo)
        cells = []
        i = 0
        while i < start:
            cell = {}
            cell['empty'] = True
            cells.append(cell)
            i = i + 1

        d = 1
        while d <= days:
            if yr<10:
                y="000" + str(yr)
            elif yr<100:
                y="00" + str(yr)
            elif yr<1000:
                y= "0" + str(yr)
            else:
                y=str(yr)
            if mo<10:
                mo_str="0" + str(mo)
            else:
                mo_str= str(mo)
            if d < 10:
                d_str ="0" + str(d)
            else:
                d_str =str(d)
            date =y + "-" + mo_str + "-" + d_str

            if date in by_date:
                cells.append(by_date[date])
            else:
                cell = {}
                cell['empty'] = True
                cell['date'] = date
                cells.append(cell)
            d = d + 1
        result[m] = cells

    return render_template("calendar.html", months_data=result, unique_months=recent)

@app.route("/day/<date>")
def day_detail(date):
    data = load_data(FILENAME)
    analysis = analyze_emotions(data)
    if date not in analysis:
        return "No records for this date", 404
    info = analysis[date]

    main = info.get('main_emotions', [])
    if len(main) == 0:
        top = info.get('dominant_emotion')
        if top:
            main = [top]
    quotes = []
    for e in main:
        if e:
            if e in info['emotion_metadata']:
                em = info['emotion_metadata'][e]
                if 'encouragement' in em:
                    lst = em['encouragement']
                    idx = random.randint(0, len(lst) - 1)
                    q = lst[idx]
                    quotes.append(q)

    return render_template("day_detail.html", date=date, data=info, encouragements=quotes)

@app.route("/weekly")
def weekly_summary():
    data = load_data(FILENAME)
    analysis = analyze_emotions(data)
    weeks = get_weekly_summary(analysis)
    return render_template("weekly_summary.html", weekly_data=weeks)

@app.route("/monthly")
def monthly_summary():
    data = load_data(FILENAME)
    analysis = analyze_emotions(data)
    months = get_monthly_summary(analysis)
    return render_template("monthly_summary.html", monthly_data=months)

if __name__ == "__main__":
    app.run(debug=True, port=5001)
```
